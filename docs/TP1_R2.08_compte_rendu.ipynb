## 2) Approximation de π : `mypi(precision)`
On utilise l'approximation :
\\[\n\\sum_{i=1}^{n} \\frac{1}{i^2} \\approx \\frac{\\pi^2}{6}\n\\Rightarrow \\pi \\approx \\sqrt{6 \\times \\sum_{i=1}^{n} \\frac{1}{i^2}}\n\\]\nOn calcule ensuite l'**erreur relative** en % par rapport à `math.pi`.


````python
import math

def mypi(precision):
    """Approximation de pi avec n termes"""
    s = 0.0
    for i in range(1, precision + 1):
        s = s + 1 / (i * i)
    return math.sqrt(6 * s)

# Test rapide (tu peux changer n)
n = 10
pi_calcule = mypi(n)
pi_vrai = math.pi
erreur = abs(pi_vrai - pi_calcule) / pi_vrai * 100

print("n =", n)
print("pi approché =", pi_calcule)
print("pi réel     =", pi_vrai)
print("erreur (%)  =", erreur)
````

### Saisie utilisateur (comme dans l'énoncé)
Cellule ci-dessous : on demande un entier `n` positif non nul, puis on affiche le résultat.


````python
import math

def mypi(precision):
    s = 0.0
    for i in range(1, precision + 1):
        s = s + 1 / (i * i)
    return math.sqrt(6 * s)

n = int(input("Entrer un entier positif (precision) : "))

pi_calcule = mypi(n)
pi_vrai = math.pi
erreur = abs(pi_vrai - pi_calcule) / pi_vrai * 100

print("pi approché =", pi_calcule)
print("pi réel     =", pi_vrai)
print("erreur relative (%) =", erreur)
````

## 4) Portée des variables (scopes)
On observe la différence entre variable **globale** et **locale**, et l'utilisation de `global`.


````python
x = "valeur globale"

def f1():
    print("--Appel f1()")
    print(" x dans f1 :", x)

def f2():
    print("--Appel f2()")
    x = "valeur locale"
    print(" x dans f2 :", x)

def f3():
    print("--Appel f3()")
    global x
    x = "valeur modifiée"
    y = 12
    print(" x dans f3 :", x)
    print(" y dans f3 :", y)

print("x au niveau global :", x)
f1()
print("x au niveau global :", x)
f2()
print("x au niveau global :", x)
f3()
print("x au niveau global :", x)

# y n'existe pas au global car créé dans f3()
# print("y au niveau global :", y)  # décommenter => NameError
````

**Commentaire :**
- `f1()` lit la variable globale `x`.
- `f2()` crée un `x` **local**, ça ne change pas le `x` global.
- `f3()` utilise `global x` pour modifier le `x` global.
- `y` est local à `f3()` : au global, l'afficher provoquerait `NameError`.

## 6) Documentation (docstrings) + `help()`
### Commentaire sur `__main__`
`if __name__ == '__main__':` exécute le code uniquement si on lance le fichier directement.
Si on **importe** le module, le code du bloc `__main__` n'est pas exécuté.

### Exemple (my_module.py)


````python
def ras():
    """Cette fonction est bien documentée mais ne fait presque rien."""
    print("rien à signaler")

if __name__ == "__main__":
    print(ras.__doc__)
    ras()
````

### Test `help(ras)`
Dans un interpréteur Python (ou une cellule Jupyter), on peut afficher la doc :


````python
def ras():
    """Cette fonction est bien documentée mais ne fait presque rien."""
    print("rien à signaler")

help(ras)
````

**Oups, corrigez ;)**
Plutôt que d'afficher seulement `ras.__doc__`, `help(ras)` donne une aide plus lisible.
